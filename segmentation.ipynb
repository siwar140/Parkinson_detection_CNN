Ce code permet de diviser un fichier audio dont la durée est supérieure à une durée donnée en plusieurs morceaux de cette durée.

```python
from IPython.display import Audio, display
import librosa

def divaudio(audio, duree):
    y, sr = librosa.load(audio, sr=None)
    samples = int(duree * sr)

    if len(y) < samples:
        print("La durée de l'audio est inférieure à la durée spécifiée.")
        return [], sr

    elif len(y) == samples:
        return [y], sr
    
    morceaux = []
    for i in range(0, len(y), samples):
        morceau = y[i:i + samples]
        if len(morceau) == samples:
            morceaux.append(morceau)
    print(f"Nombre de morceaux créés : {len(morceaux)}")
    return morceaux, sr


audio = r"C:\Users\MSI\Desktop\Parkinson_VF\DATA\Original\Patologicas\AVPEPUDEA0006i1.wav"
morceaux, sr = divaudio(audio, 1)
y, _ = librosa.load(audio, sr=None)
duree_audio = len(y) / sr

print(f"Durée de l'audio : {duree_audio:.2f} s")
display(Audio(data=audio, rate=sr))
for i in range(len(morceaux)):
    print(f"Durée du morceau {i + 1} : {len(morceaux[i]) / sr:.2f} s")
    display(Audio(data=morceaux[i], rate=sr))
```

```python
import os
import librosa
import soundfile as sf

BD_path = "Original"
duration = 3
classe = ["Control", "Patologicas"]
for c in classe:
    class_path = os.path.join(BD_path, c)
    for file in os.listdir(class_path):
        if file.endswith(".wav"):
            audio_path = os.path.join(class_path, file)
            morceaux, sr = divaudio(audio_path, duration)  # Segmentation en morceaux de `duration` secondes
            for i, morceau in enumerate(morceaux):
                output_dir = os.path.join(f"Segmented_{duration}", c)
                os.makedirs(output_dir, exist_ok=True)
                output_file = os.path.join(output_dir, f"{os.path.splitext(file)[0]}_segment_{i + 1}.wav")
                sf.write(output_file, morceau, sr)
```

```python
#Ce code 

from pathlib import Path
import shutil
import librosa
import soundfile as sf




workspace_root = Path.cwd()
source = workspace_root / "Original"
duration_sec = 3
destination = workspace_root / f"{source.name}_{duration_sec}s"

if destination.exists():
    shutil.rmtree(destination)
destination.mkdir(parents=True, exist_ok=True)

wave_files = sorted(source.rglob("*.wav"))
created_files = 0

for audio_path in wave_files:
    relative_dir = audio_path.parent.relative_to(source)
    output_dir = destination / relative_dir
    output_dir.mkdir(parents=True, exist_ok=True)

    chunks, sr = divaudio(str(audio_path), duration_sec)
    stem = audio_path.stem

    if not chunks:
        print(f"Ignoré : {audio_path} ({len(librosa.load(audio_path, sr=None)[0]) / librosa.load(audio_path, sr=None)[1]:.2f}s)")
        continue

    for idx, chunk in enumerate(chunks, start=1):
        output_file = output_dir / f"{stem}_{idx:02d}.wav"
        sf.write(output_file, chunk, sr)
        created_files += 1

print(f"Dataset créé dans : {destination}")
print(f"Fichiers WAV traités : {len(wave_files)}")
print(f"Segments audio écrits : {created_files}")
```
